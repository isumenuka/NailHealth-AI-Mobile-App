## 8️⃣ Adapting the AI for Dermatology (Specialized Fine-Tuning)

````python
class MedSigLIPClassifier(nn.Module):
    def __init__(self, medsiglip_model, num_classes, device0='cuda:0', device1='cuda:1'):
        super().__init__()
        self.medsiglip = medsiglip_model.to(device0)
        self.device0 = device0
        self.device1 = device1

        embed_dim = 1152

        self.classifier = nn.Sequential(
            nn.Linear(embed_dim, 768),
            nn.LayerNorm(768),
            nn.GELU(),
            nn.Dropout(0.5),
            
            nn.Linear(768, 512),
            nn.LayerNorm(512),
            nn.GELU(),
            nn.Dropout(0.5),
            
            nn.Linear(512, 256),
            nn.LayerNorm(256),
            nn.GELU(),
            nn.Dropout(0.4),
            
            nn.Linear(256, num_classes)
        ).to(device1)

        # FREEZE ALL MedSigLIP layers to save memory
        for param in self.medsiglip.parameters():
            param.requires_grad = False

    def forward(self, images):
        # Images on GPU 0
        images = images.to(self.device0)
        
        # Get embeddings on GPU 0
        with torch.no_grad():
            outputs = self.medsiglip.vision_model(pixel_values=images)
            embeddings = outputs.pooler_output
        
        # Move embeddings to GPU 1 and cast to FP32
        embeddings = embeddings.to(self.device1).float()
        
        # Classifier on GPU 1
        logits = self.classifier(embeddings)
        return logits


num_classes = len(train_dataset.classes)

# Check available GPUs
num_gpus = torch.cuda.device_count()
if num_gpus >= 2:
    print(f"🚀 Using 2 GPUs for Model Parallel!")
    device0 = 'cuda:0'
    device1 = 'cuda:1'
    classifier = MedSigLIPClassifier(
        medsiglip_model=model,
        num_classes=num_classes,
        device0=device0,
        device1=device1
    )
else:
    print(f"⚠️ Only {num_gpus} GPU(s) detected. Using single GPU.")
    device0 = 'cuda:0'
    classifier = MedSigLIPClassifier(
        medsiglip_model=model,
        num_classes=num_classes,
        device0=device0,
        device1=device0
    )

print(f"✅ Classifier ready! Classes: {num_classes}")
print(f"   MedSigLIP (Feature Extractor): {device0}")
print(f"   Classification Head: {device1 if num_gpus >= 2 else device0}")
print(f"   Optimized classifier: 1152→768→512→256→{num_classes}")
````

## 8½. Automated Hyperparameter Tuning

````python

# --- MANDATORY AUTOMATED HYPERPARAMETER TUNING ---
# We use Optuna to find the optimal Learning Rate, Batch Size, and Optimizer.
# This ensures the model is perfectly tuned for your specific dataset version.


N_TRIALS = 10  # Number of trials to run
BEST_LR = None
BEST_BATCH_SIZE = None
BEST_OPTIMIZER = None

def objective(trial):
    # 1. Suggest Hyperparameters
    lr = trial.suggest_float("lr", 1e-5, 1e-3, log=True)
    batch_size = trial.suggest_categorical("batch_size", [16, 32])
    optimizer_name = trial.suggest_categorical("optimizer", ["AdamW", "SGD"])
    
    # 2. Setup Data Loaders (Re-init with new batch size)
    train_loader_tune = DataLoader(train_dataset, batch_size=batch_size, shuffle=True, pin_memory=True, num_workers=0)
    test_loader_tune = DataLoader(test_dataset, batch_size=batch_size, shuffle=False, pin_memory=True, num_workers=0)
    
    # 3. Initialize Model (Fresh start)
    # Re-instantiate classifier head only
    classifier_tune = MedSigLIPClassifier(model, num_classes, device0='cuda:0', device1='cuda:0' if torch.cuda.device_count() == 1 else 'cuda:1')
    
    # 4. Setup Optimizer
    params = list(classifier_tune.classifier.parameters())
    if optimizer_name == "AdamW":
        optimizer_tune = optim.AdamW(params, lr=lr, weight_decay=1e-3)
    else:
        optimizer_tune = optim.SGD(params, lr=lr, momentum=0.9, weight_decay=1e-3)
        
    # 5. Training Loop (Short - 3 Epochs)
    criterion_tune = nn.CrossEntropyLoss()
    
    classifier_tune.train()
    for epoch in range(3): 
        for images, labels in train_loader_tune:
            images = images.to(classifier_tune.device0)
            labels = labels.to(classifier_tune.device1)
            
            optimizer_tune.zero_grad()
            outputs = classifier_tune(images)
            loss = criterion_tune(outputs, labels)
            loss.backward()
            optimizer_tune.step()
            
        # Validation
        classifier_tune.eval()
        correct = 0
        total = 0
        with torch.no_grad():
            for images, labels in test_loader_tune:
                images = images.to(classifier_tune.device0)
                labels = labels.to(classifier_tune.device1)
                outputs = classifier_tune(images)
                _, predicted = torch.max(outputs.data, 1)
                total += labels.size(0)
                correct += (predicted == labels).sum().item()
        
        accuracy = correct / total
        
        # Pruning
        trial.report(accuracy, epoch)
        if trial.should_prune():
            raise optuna.exceptions.TrialPruned()
            
    return accuracy

print("🔬 STARTING AUTOMATED HYPERPARAMETER TUNING...")
print(f"   Running {N_TRIALS} trials to find optimal settings...")

study = optuna.create_study(direction="maximize")
try:
    study.optimize(objective, n_trials=N_TRIALS)
    
    print("🎉 Tuning Complete!")
    print("   Best Trial:")
    print(f"   Accuracy: {study.best_trial.value:.4f}")
    print("   Params: ")
    for key, value in study.best_trial.params.items():
        print(f"      {key}: {value}")
        
    # SAVE BEST PARAMS GLOBALLY
    BEST_LR = study.best_trial.params['lr']
    BEST_BATCH_SIZE = study.best_trial.params['batch_size']
    BEST_OPTIMIZER = study.best_trial.params['optimizer']
    
except Exception as e:
    print(f"⚠️ Tuning failed or interrupted: {e}")
    print("   Using default parameters.")

````

## 9️⃣ Configuring the Learning Process

````python

NUM_EPOCHS = 10
LEARNING_RATE = 1e-3  # Increased since we're only training classifier
WEIGHT_DECAY = 1e-3

# --- APPLY OPTUNA RESULTS ---
if 'BEST_LR' in globals() and BEST_LR is not None:
    print("\n🚀 APPLYING TUNED HYPERPARAMETERS:")
    print(f"   Overriding LR: {LEARNING_RATE} -> {BEST_LR}")
    LEARNING_RATE = BEST_LR
    print(f"   Overriding Batch Size: {BATCH_SIZE} -> {BEST_BATCH_SIZE}")
    BATCH_SIZE = BEST_BATCH_SIZE
    # Note: Batch Size update requires re-creating DataLoaders if we want it strictly applied,
    # but for simplicity we might just stick to LR/Optimizer or re-init dataloaders here.
    # Let's re-init dataloaders just to be safe!
    train_loader = DataLoader(train_dataset, batch_size=BATCH_SIZE, shuffle=True, pin_memory=True, num_workers=NUM_WORKERS)
    test_loader = DataLoader(test_dataset, batch_size=BATCH_SIZE, shuffle=False, pin_memory=True, num_workers=NUM_WORKERS)
else:
    print("\nℹ️ Using default hyperparameters (Tuning skipped or failed).")
GRADIENT_ACCUMULATION_STEPS = 1

# --- BALANCED SAMPLING: Calculate Class Weights ---
print("⚖️ Calculating class weights for balanced loss...")
try:
    # Get all targets (labels) from the training dataset
    if hasattr(train_dataset, 'targets'):
        targets = train_dataset.targets
        class_counts = np.bincount(targets)
    else:
        # Fallback for datasets that don't expose targets directly 
        targets = [y for _, y in train_dataset]
        class_counts = np.bincount(targets)
    
    total_samples = len(targets)
    num_classes = len(class_counts)
    
    # Calculate weights: w_j = n_samples / (n_classes * n_samples_j)
    # Add small epsilon to avoid division by zero if a class has 0 samples (unlikely but safe)
    class_weights = total_samples / (num_classes * (class_counts + 1e-6))
    
    # Convert to tensor
    class_weights_tensor = torch.FloatTensor(class_weights).to(device)
    
    print(f"   📊 Class Counts: {class_counts}")
    print(f"   ⚖️ Class Weights: {np.round(class_weights, 4)}")
    
    # Update criterion with weights
    criterion = nn.CrossEntropyLoss(weight=class_weights_tensor, label_smoothing=0.1)
    print("✅ Class-weighted CrossEntropyLoss activated!")

except Exception as e:
    print(f"⚠️ Could not calculate class weights: {e}")
    print("   Using standard CrossEntropyLoss instead.")
    criterion = nn.CrossEntropyLoss(label_smoothing=0.1)
# --------------------------------------------------

# Only optimize classifier (MedSigLIP is frozen)
classifier_params = list(classifier.classifier.parameters())


if 'BEST_OPTIMIZER' in globals() and BEST_OPTIMIZER == 'SGD':
    print("   Using Optimzer: SGD")
    optimizer = optim.SGD(classifier_params, lr=LEARNING_RATE, momentum=0.9, weight_decay=WEIGHT_DECAY)
else:
    print("   Using Optimizer: AdamW")
    optimizer = optim.AdamW(classifier_params, lr=LEARNING_RATE, weight_decay=WEIGHT_DECAY, betas=(0.9, 0.999))

total_steps = len(train_loader) * NUM_EPOCHS // GRADIENT_ACCUMULATION_STEPS
scheduler = ReduceLROnPlateau(
    optimizer,
    mode='min',
    factor=0.1,
    patience=2,
    verbose=True
)

print("✅ Advanced training configuration (Classifier-Only Fine-Tuning):")
print(f"   📍 Epochs: {NUM_EPOCHS}")
print(f"   📍 Learning Rate: {LEARNING_RATE}")
print(f"   📍 Batch Size: {BATCH_SIZE}")
print(f"   📍 Gradient Accumulation: {GRADIENT_ACCUMULATION_STEPS}")
print(f"   📍 Optimizer: {optimizer.__class__.__name__} with ReduceLROnPlateau")
print(f"   📍 Weight Decay: {WEIGHT_DECAY}")
print(f"   📍 Label Smoothing: 0.1")
print(f"   📍 MedSigLIP: FROZEN (memory efficient)")
print(f"   📍 GPU Strategy: Model Parallel (GPU 0 + GPU 1)")

# --- MIXED PRECISION TRAINING ---
scaler = GradScaler()
print('⚡ Mixed Precision (GradScaler) activated!')
# --- ADVANCED AUGMENTATION: Mixup & Cutmix ---
MIXUP_ARGS = {
    'mixup_alpha': 0.8,
    'cutmix_alpha': 1.0,
    'cutmix_minmax': None,
    'prob': 1.0,
    'switch_prob': 0.5,
    'mode': 'batch',
    'label_smoothing': 0.1,
    'num_classes': num_classes
}

mixup_fn = Mixup(**MIXUP_ARGS)
criterion_soft = SoftTargetCrossEntropy()
print('🎨 Mixup & CutMix augmentation activated!')
````

````python

def train_epoch(model, train_loader, criterion, optimizer, scheduler, accumulation_steps=1, mixup_fn=None):
    model.train()
    total_loss = 0
    all_preds = []
    all_labels = []
    optimizer.zero_grad()
    
    # Determine devices
    input_device = 'cuda:0' if torch.cuda.device_count() > 0 else 'cpu'
    # For model parallel, classification often happens on the last device (e.g., cuda:1)
    target_device = 'cuda:1' if torch.cuda.device_count() >= 2 else input_device

    pbar = tqdm(train_loader, desc="Learning (w/ Mixup)")
    for step, (images, labels) in enumerate(pbar):
        images = images.to(input_device)
        labels = labels.to(input_device)

        if mixup_fn is not None:
            images, labels = mixup_fn(images, labels)
        
        # If labels are tensor (mixup produces tensor), move to target device if needed
        # Note: if mixup is OFF, labels is tensor [B]. If ON, labels is tensor [B, C].
        if isinstance(labels, torch.Tensor) and labels.device != torch.device(target_device):
            labels = labels.to(target_device)

        with autocast():
            outputs = model(images)
        
        loss = criterion(outputs, labels)
        
        if math.isnan(loss.item()):
            print(f"   ⚠️ NaN loss at step {step}, skipping")
            optimizer.zero_grad()
            continue
            
        loss = loss / accumulation_steps
        scaler.scale(loss).backward()
        
        if (step + 1) % accumulation_steps == 0:
            scaler.unscale_(optimizer)
            torch.nn.utils.clip_grad_norm_(model.classifier.parameters(), max_norm=1.0)
            scaler.step(optimizer)
            scaler.update()
            # Note: scheduler.step() removed for ReduceLROnPlateau
            optimizer.zero_grad()

        total_loss += loss.item() * accumulation_steps
        
        preds = outputs.argmax(dim=1)
        
        # Recover hard labels for accuracy calc
        if len(labels.shape) > 1: 
             logging_labels = labels.argmax(dim=1) 
        else:
             logging_labels = labels

        all_preds.extend(preds.detach().cpu().numpy())
        all_labels.extend(logging_labels.detach().cpu().numpy())

        pbar.set_postfix({'loss': f'{loss.item()*accumulation_steps:.4f}'})

    avg_loss = total_loss / len(train_loader)
    accuracy = accuracy_score(all_labels, all_preds) if len(all_labels) > 0 else 0
    return avg_loss, accuracy
````

## 1️⃣1️⃣ Run ADVANCED Training (10 Epochs)

````python
history = {
    'train_loss': [], 'train_acc': [], 'test_loss': [], 'test_acc': [],
    'test_precision': [], 'test_recall': [], 'test_f1': [], 'learning_rate': []
}

best_accuracy = 0
best_epoch = 0
patience_counter = 0
max_patience = 5
best_model_path = os.path.join(OUTPUT_PATH, 'best_model.pt')

best_val_loss = float('inf')
patience_counter = 0
EARLY_STOPPING_PATIENCE = 5

print("\n" + "="*70)
print("🚀 COMMENCING MEDICAL AI TRAINING")
print(f"🚀 Using {torch.cuda.device_count()} GPUs (Model Parallel)")
print("="*70)

for epoch in range(NUM_EPOCHS):
    print(f"\n📊 Epoch {epoch+1}/{NUM_EPOCHS}")

    train_loss, train_acc = train_epoch(classifier, train_loader, criterion_soft, optimizer, scheduler, GRADIENT_ACCUMULATION_STEPS, mixup_fn=mixup_fn)
    history['train_loss'].append(train_loss)
    history['train_acc'].append(train_acc)
    history['learning_rate'].append(optimizer.param_groups[0]['lr'])

    test_loss, test_acc, test_prec, test_rec, test_f1, preds, labels = evaluate(classifier, test_loader, criterion)
    history['test_loss'].append(test_loss)
    history['test_acc'].append(test_acc)
    history['test_precision'].append(test_prec)
    history['test_recall'].append(test_rec)
    history['test_f1'].append(test_f1)
    scheduler.step(test_loss)

    # Early Stopping (Monitor Loss)
    if test_loss < best_val_loss:
        best_val_loss = test_loss
        patience_counter = 0
        # Checkpoint for best loss could be saved here too, but we are saving best acc already.
    else:
        patience_counter += 1
        print(f"   ⏳ No loss improvement for {patience_counter}/{EARLY_STOPPING_PATIENCE} epochs.")
        if patience_counter >= EARLY_STOPPING_PATIENCE:
            print(f"   🛑 Early stopping triggered!")
            break

    print(f"   LR: {optimizer.param_groups[0]['lr']:.8f}")

    print(f"   Train Loss: {train_loss:.4f} | Train Acc: {train_acc:.4f}")
    print(f"   Test Loss: {test_loss:.4f} | Test Acc: {test_acc:.4f}")
    print(f"   Precision: {test_prec:.4f} | Recall: {test_rec:.4f} | F1: {test_f1:.4f}")
    print(f"   LR: {optimizer.param_groups[0]['lr']:.6f}")

    if test_acc > best_accuracy:
        best_accuracy = test_acc
        best_epoch = epoch + 1
        torch.save(classifier.state_dict(), best_model_path)
        patience_counter = 0
        print(f"   ⭐ BEST model saved! (Accuracy: {best_accuracy:.4f})")
    else:
        patience_counter += 1

    torch.cuda.empty_cache()
    gc.collect()

print("\n" + "="*70)
print("✅ TRAINING COMPLETED")
print(f"   Best Accuracy: {best_accuracy:.4f} at Epoch {best_epoch}")
print("="*70)

history_path = os.path.join(OUTPUT_PATH, 'training_history.json')
with open(history_path, 'w') as f:
    json.dump(history, f, indent=4)
print(f"\n💾 Training history saved to: {history_path}")
````

## 1️⃣1️⃣B - 🧠 Stage 2: Fine-Tuning (Gradual Unfreezing)

Now that the classifier head is trained, we slightly unfreeze the last layers of the MedSigLIP vision encoder to adapt the feature extraction specifically for nail textures.

````python

# 1. Configuration — defined FIRST to avoid NameError
UNFREEZE_LAYERS = 2  # Number of last encoder layers to unfreeze

print("\n" + "="*70)
print("🚀 COMMENCING STAGE 2: FINE-TUNING DEEPER LAYERS")
print(f"   (Unfreezing last {UNFREEZE_LAYERS} layers of vision encoder)")
print("="*70)

# 1. Unfreeze Last Layers
print(f"🔓 Unfreezing last {UNFREEZE_LAYERS} layers...")

# Access the vision model (MedSigLIP)
# Structure: model -> medsiglip -> vision_model -> encoder -> layers
try:
    # Unfreeze the last N layers of the encoder
    for layer in classifier.medsiglip.vision_model.encoder.layers[-UNFREEZE_LAYERS:]:
        for param in layer.parameters():
            param.requires_grad = True
    
    # Also unfreeze the post-layernorm if present (usually good practice)
    for param in classifier.medsiglip.vision_model.post_layernorm.parameters():
        param.requires_grad = True
        
    print("✅ Layers unfrozen successfully.")
    
except Exception as e:
    print(f"⚠️ Error unfreezing layers: {e}")
    print("   Continuing with only classifier training...")

# 2. Re-initialize Optimizer with Lower Learning Rate
FINE_TUNE_LR = 1e-4  # 10x smaller than initial LR
FINE_TUNE_EPOCHS = 5
print(f"📉 Setting fine-tuning LR to {FINE_TUNE_LR} for {FINE_TUNE_EPOCHS} epochs")

# Get all trainable parameters (classifier + unfrozen layers)
trainable_params = filter(lambda p: p.requires_grad, classifier.parameters())

optimizer_ft = optim.AdamW(trainable_params, lr=FINE_TUNE_LR, weight_decay=WEIGHT_DECAY*2)

total_steps_ft = len(train_loader) * FINE_TUNE_EPOCHS // GRADIENT_ACCUMULATION_STEPS
scheduler_ft = ReduceLROnPlateau(
    optimizer_ft,
    mode='min',
    factor=0.1,
    patience=2,
    verbose=True
)

# 3. Fine-tuning Loop
best_accuracy_ft = best_accuracy # Start tracking from previous best
best_epoch_ft = best_epoch
best_val_loss_ft = float('inf')
patience_counter_ft = 0
EARLY_STOPPING_PATIENCE_FT = 3


for epoch in range(NUM_EPOCHS, NUM_EPOCHS + FINE_TUNE_EPOCHS):
    print(f"\n📊 Fine-Tuning Epoch {epoch+1}/{NUM_EPOCHS + FINE_TUNE_EPOCHS}")

    # Use the NEW optimizer and scheduler
    train_loss, train_acc = train_epoch(classifier, train_loader, criterion_soft, optimizer_ft, scheduler_ft, GRADIENT_ACCUMULATION_STEPS, mixup_fn=mixup_fn)
    history['train_loss'].append(train_loss)
    history['train_acc'].append(train_acc)
    history['learning_rate'].append(optimizer_ft.param_groups[0]['lr'])

    test_loss, test_acc, test_prec, test_rec, test_f1, preds, labels = evaluate(classifier, test_loader, criterion)
    history['test_loss'].append(test_loss)
    history['test_acc'].append(test_acc)
    history['test_precision'].append(test_prec)
    history['test_recall'].append(test_rec)
    history['test_f1'].append(test_f1)
    scheduler_ft.step(test_loss)
    # Early Stopping (Monitor Loss)
    if test_loss < best_val_loss_ft:
        best_val_loss_ft = test_loss
        patience_counter_ft = 0
    else:
        patience_counter_ft += 1
        print(f"   ⏳ No loss improvement for {patience_counter_ft}/{EARLY_STOPPING_PATIENCE_FT} epochs.")
        if patience_counter_ft >= EARLY_STOPPING_PATIENCE_FT:
            print(f"   🛑 Fine-tuning Early stopping triggered!")
            break


    print(f"   Train Loss: {train_loss:.4f} | Train Acc: {train_acc:.4f}")
    print(f"   Test Loss: {test_loss:.4f} | Test Acc: {test_acc:.4f}")
    print(f"   Precision: {test_prec:.4f} | Recall: {test_rec:.4f} | F1: {test_f1:.4f}")
    print(f"   LR: {optimizer_ft.param_groups[0]['lr']:.6f}")

    if test_acc > best_accuracy_ft:
        best_accuracy_ft = test_acc
        best_epoch_ft = epoch + 1
        torch.save(classifier.state_dict(), best_model_path)
        print(f"   ⭐ BEST fine-tuned model saved! (Accuracy: {best_accuracy_ft:.4f})")
    
    torch.cuda.empty_cache()
    gc.collect()

# Update global best for final reporting
best_accuracy = max(best_accuracy, best_accuracy_ft)
if best_accuracy_ft > best_accuracy:
    best_epoch = best_epoch_ft

print("\n" + "="*70)
print("✅ FINE-TUNING COMPLETED")
print(f"   Final Best Accuracy: {best_accuracy:.4f}")
print("="*70)
````

````python

# ─────────────────────────────────────────────────────────────────────────────
# STEP 13 — Build the HuggingFace-Inference-Endpoints-compatible model package
# ─────────────────────────────────────────────────────────────────────────────
DEPLOYMENT_PATH = os.path.join(OUTPUT_PATH, 'medsiglip_nail_classifier_hf')
os.makedirs(DEPLOYMENT_PATH, exist_ok=True)

print("="*70)
print("🚀 PREPARING MODEL FOR HUGGINGFACE INFERENCE ENDPOINTS")
print("="*70)

# ── Step 1: load best weights ──────────────────────────────────────────────
print("\n📥 Step 1: Loading best trained model...")
classifier.load_state_dict(torch.load(best_model_path, map_location='cpu'))
classifier.eval()
print("✅ Best model loaded")

# ── Step 2: save weights ───────────────────────────────────────────────────
print("\n💾 Step 2: Saving model weights...")
checkpoint = {
    'classifier_state_dict': classifier.classifier.state_dict(),
    'full_state_dict':       classifier.state_dict(),
    'num_classes':           num_classes,
    'class_names':           train_dataset.classes,
    'image_size':            IMAGE_SIZE,
    'best_accuracy':         float(best_accuracy),
    'best_epoch':            int(best_epoch),
}
torch.save(checkpoint, os.path.join(DEPLOYMENT_PATH, 'pytorch_model.bin'))
print("✅ Weights saved → pytorch_model.bin")

# ── Step 3: save processor ────────────────────────────────────────────────
print("\n💾 Step 3: Saving processor...")
processor.save_pretrained(DEPLOYMENT_PATH)
print("✅ Processor saved")

# ── Step 4: config.json  ← KEY FIX: model_type = "custom" NOT "medsiglip-classifier"
print("\n📝 Step 4: Writing config.json...")
config_data = {
    "model_type":             "custom",
    "base_model":             "google/medsiglip-448",
    "num_classes":            num_classes,
    "class_names":            train_dataset.classes,
    "image_size":             IMAGE_SIZE,
    "embedding_dim":          1152,
    "classifier_hidden_dims": [768, 512, 256],
    "best_accuracy":          float(best_accuracy),
    "best_epoch":             int(best_epoch),
    "framework":              "pytorch",
    "task":                   "image-classification",
}
with open(os.path.join(DEPLOYMENT_PATH, 'config.json'), 'w') as f:
    json.dump(config_data, f, indent=2)
print("✅ config.json written  (model_type='custom')")

# ── Step 5: handler.py ────────────────────────────────────────────────────
print("\n🔧 Step 5: Writing handler.py...")
handler_code = '''"""
Custom EndpointHandler for MedSigLIP Nail Disease Classifier.
Bypasses AutoConfig so HF Inference Endpoints can serve this custom architecture.
Set env-var HUGGING_FACE_HUB_TOKEN on the endpoint to access google/medsiglip-448.
"""
import os, json, base64, logging
from io import BytesIO
from typing import Dict, Any, List

import torch
import torch.nn as nn
from PIL import Image
from transformers import AutoModel, AutoProcessor

logger = logging.getLogger(__name__)


def build_classifier_head(embed_dim: int, num_classes: int) -> nn.Sequential:
    """Recreate the exact same classifier head used during training."""
    return nn.Sequential(
        nn.Linear(embed_dim, 768), nn.LayerNorm(768), nn.GELU(), nn.Dropout(0.5),
        nn.Linear(768, 512),       nn.LayerNorm(512), nn.GELU(), nn.Dropout(0.5),  # ✅ FIXED: matches training
        nn.Linear(512, 256),       nn.LayerNorm(256), nn.GELU(), nn.Dropout(0.4),  # ✅ FIXED: matches training
        nn.Linear(256, num_classes),
    )


class MedSigLIPClassifierModel(nn.Module):
    def __init__(self, base_model: nn.Module, head: nn.Sequential):
        super().__init__()
        self.vision_model = base_model.vision_model
        self.classifier   = head

    def forward(self, pixel_values: torch.Tensor) -> torch.Tensor:
        with torch.no_grad():
            outputs    = self.vision_model(pixel_values=pixel_values)
            embeddings = outputs.pooler_output
        return self.classifier(embeddings.float())


class EndpointHandler:
    """HuggingFace Inference Endpoints entry-point."""

    def __init__(self, path: str = ""):
        logger.info(f"[handler] Loading model from {path!r} ...")

        # load config
        with open(os.path.join(path, "config.json")) as f:
            cfg = json.load(f)
        self.class_names = cfg["class_names"]
        num_classes      = cfg["num_classes"]
        base_model_id    = cfg.get("base_model", "google/medsiglip-448")
        logger.info(f"[handler] Classes ({num_classes}): {self.class_names}")

        # load base model (HUGGING_FACE_HUB_TOKEN env-var must be set)
        logger.info(f"[handler] Downloading base model: {base_model_id}")
        base = AutoModel.from_pretrained(base_model_id, torch_dtype=torch.float32)

        # load processor
        self.processor = AutoProcessor.from_pretrained(path)

        # build & restore model
        head        = build_classifier_head(1152, num_classes)
        self.model  = MedSigLIPClassifierModel(base, head)

        ckpt = torch.load(os.path.join(path, "pytorch_model.bin"), map_location="cpu")
        if "full_state_dict" in ckpt:
            self.model.load_state_dict(ckpt["full_state_dict"], strict=False)
        elif "classifier_state_dict" in ckpt:
            self.model.classifier.load_state_dict(ckpt["classifier_state_dict"])
        elif "model_state_dict" in ckpt:
            self.model.load_state_dict(ckpt["model_state_dict"], strict=False)
        else:
            raise ValueError("pytorch_model.bin: no recognised state_dict key")

        self.model.eval()
        logger.info("[handler] Model ready ✅")

    def __call__(self, data: Dict[str, Any]) -> List[Dict[str, Any]]:
        """
        Input:  {"inputs": "<base64-encoded-image>"}
        Output: [{"label": "psoriasis", "score": 0.72}, ...]
        """
        raw = data.get("inputs") or data.get("image")

        if isinstance(raw, str):
            image = Image.open(BytesIO(base64.b64decode(raw))).convert("RGB")
        elif isinstance(raw, Image.Image):
            image = raw.convert("RGB")
        else:
            raise ValueError(f"Unsupported input type: {type(raw)}")

        # processor reads image_size from preprocessor_config.json — no need to pass size=
        encoded      = self.processor(images=image, return_tensors="pt")
        pixel_values = encoded["pixel_values"]

        with torch.no_grad():
            probs = torch.softmax(self.model(pixel_values), dim=-1)[0]

        results = [
            {"label": self.class_names[i], "score": round(probs[i].item(), 6)}
            for i in range(len(self.class_names))
        ]
        results.sort(key=lambda x: x["score"], reverse=True)
        return results
'''

with open(os.path.join(DEPLOYMENT_PATH, 'handler.py'), 'w') as f:
    f.write(handler_code)
print("✅ handler.py written")

# ── Step 6: requirements.txt ──────────────────────────────────────────────
print("\n📦 Step 6: Writing requirements.txt...")
with open(os.path.join(DEPLOYMENT_PATH, 'requirements.txt'), 'w') as f:
    f.write("transformers>=4.40.0\ntorch>=2.0.0\nPillow>=9.0.0\ntimm>=0.9.0\n")
print("✅ requirements.txt written")

# ── Step 7: README ────────────────────────────────────────────────────────
print("\n📄 Step 7: Writing README.md...")
readme_content = f"""---
license: apache-2.0
tags:
- medical
- image-classification
- nail-disease
- medsiglip
library_name: generic
pipeline_tag: image-classification
---

# MedSigLIP Nail Disease Classifier

Fine-tuned from [google/medsiglip-448](https://huggingface.co/google/medsiglip-448).
Detects **{{num_classes}} nail conditions** with **{{best_accuracy*100:.2f}}%** accuracy.

## Classes
{{chr(10).join([f"- `{{c}}`" for c in train_dataset.classes])}}

## Deploying via HuggingFace Inference Endpoints

This repo uses a custom `handler.py` — no AutoConfig needed.

### Required env-var on the endpoint
```
HUGGING_FACE_HUB_TOKEN = hf_xxxxxxxxxxxx
```
(Needed to download the gated base model `google/medsiglip-448` at container startup.)

### Input format
```json
{{"inputs": "<BASE64_IMAGE_STRING>"}}
```

### Output format
```json
[
  {{"label": "psoriasis",  "score": 0.72}},
  {{"label": "pitting",    "score": 0.15}}
]
```

## Disclaimer
Research use only. Not for clinical diagnosis.
"""
with open(os.path.join(DEPLOYMENT_PATH, 'README.md'), 'w') as f:
    f.write(readme_content)
print("✅ README.md written")

# Summary
print("\n" + "="*70)
print("✅ DEPLOYMENT PACKAGE READY")
print("="*70)
print(f"\n📦 Package Contents:")
for item in sorted(os.listdir(DEPLOYMENT_PATH)):
    fp = os.path.join(DEPLOYMENT_PATH, item)
    if os.path.isfile(fp):
        print(f"   • {{item}}  ({{os.path.getsize(fp)/1024/1024:.2f}} MB)")
    else:
        print(f"   • {{item}}/")
print("\n📤 Run the Upload cell below to push to HuggingFace Hub")
````
